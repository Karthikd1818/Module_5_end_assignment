### Task 2 – CNN Model Development

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Conv2D,
    MaxPooling2D,
    Flatten,
    Dense,
    Dropout
)

print("TensorFlow:", tf.__version__)
```

```python
data = np.load("cifar10_preprocessed.npz")

x_train = data["x_train"]
y_train = data["y_train"]

x_val = data["x_val"]
y_val = data["y_val"]

x_test = data["x_test"]
y_test = data["y_test"]
```

### Create the CNN model

```python
model = Sequential([

    Conv2D(32, (3, 3), activation='relu',
           padding='same', input_shape=(32, 32, 3)),

    MaxPooling2D((2, 2)),

    Conv2D(64, (3, 3), activation='relu',
           padding='same'),

    MaxPooling2D((2, 2)),

    Flatten(),

    Dense(128, activation='relu'),

    Dropout(0.5),

    Dense(10, activation='softmax')
])
```

```python
model.summary()
```

```python
print("Model Created Successfully!")

print("Input Shape:", model.input_shape)
print("Output Shape:", model.output_shape)

print("Total Parameters:",
      model.count_params())
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    x_train,
    y_train,
    validation_data=(x_val, y_val),
    epochs=10,
    batch_size=32
)
```

```python
cnn_test_loss, cnn_test_accuracy = model.evaluate(
    x_test,
    y_test
)

print("CNN Test Loss:", cnn_test_loss)
print("CNN Test Accuracy:", cnn_test_accuracy * 100)
```

```python
model.save("custom_cnn_model.keras")
```
